```python
from pathlib import Path

import pandas as pd
from ydata_profiling import ProfileReport

# Paths are resolved from the notebook's folder.
input_path = Path(r"..\3. Feature Engineering\gurgaon_properties_cleaned_v2.csv")
output_path = Path("output_report.html")

if not input_path.exists():
    raise FileNotFoundError(f"Dataset not found: {input_path.resolve()}")

df = pd.read_csv(input_path).drop_duplicates()
profile = ProfileReport(df, title="Gurgaon Properties Profiling Report", explorative=True)
profile.to_file(output_path)

print(f"Profile report written to: {output_path.resolve()}")
```
